```python
import physics.symbolic as s
import physics.functions as p
import physics.notebook as nb
import numpy as np
import scipy.constants as spc
import astropy.constants.iau2015 as ac
import matplotlib.pyplot as plt
nb.enablePrintAll()
```

```python
A = s.tau * s.v**2 / smp.sqrt(s.c**2 - s.v**2)
A
```

```python
integrated = A.subs([[s.tau, s.v / s.vbar]]).integrate(s.v)
integrated.simplify()
```

```python
vels = np.linspace(0, spc.c, 1000)
f = smp.lambdify([s.v], integrated.subs([[s.vbar, s.vbarCosmological().subs(s.subsAll)], [s.c, spc.c], [s.tau, 1]]))
fig, ax = plt.subplots()
ax.plot(vels, f(vels))
```

```python
f(0)
f(p.vbarCosmological().value)
```

```python
dy = A
dv = s.v * s.tau
dy_dv = dy / dv
dy_dv
```

```python
dy_dv.subs([[s.v, s.vbarCosmological()], *s.subsAll])
```

```python
f = s.beta * s.x - dy_dv
sols = smp.solve(f, s.x)
sols
```
